## 5. Preprocessamento (Pipelines e Transformadores)

Nesta fase, utilizamos o `ColumnTransformer` para aplicar diferentes pré-processamentos a diferentes tipos de colunas de forma automatizada e robusta, evitando o data leakage.
- **Numéricas:** Imputação de valores em falta (se existirem) e Escalamento Standard.
- **Categóricas:** Imputação e One-Hot Encoding.

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split

# Identificação de colunas
target_reg = 'MonthlyIncome'
target_clf = 'Attrition'

# Colunas a remover
drop_features = ['EmployeeCount', 'StandardHours', 'Over18', 'EmployeeNumber']
df_clean = df.drop(columns=drop_features)

# Definir features numéricas e categóricas
numeric_features = df_clean.select_dtypes(include=['int64', 'float64']).drop(columns=[target_reg]).columns.tolist()
categorical_features = df_clean.select_dtypes(include=['object']).drop(columns=[target_clf]).columns.tolist()

# Transformers
numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='constant', fill_value='missing')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])

preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_features),
        ('cat', categorical_transformer, categorical_features)
    ])
```

## 6. Problema de Regressão: Prever MonthlyIncome

Utilizaremos o RandomForestRegressor para prever o MonthlyIncome.

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import root_mean_squared_error

X_reg = df_clean.drop(columns=[target_reg, target_clf])
y_reg = df_clean[target_reg]

X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(X_reg, y_reg, test_size=0.2, random_state=42)

# Definir e treinar o modelo Random Forest Regressor
rf_reg = RandomForestRegressor(random_state=42)
pipe_reg = Pipeline(steps=[('preprocessor', preprocessor), ('regressor', rf_reg)])
pipe_reg.fit(X_train_reg, y_train_reg)
y_pred_reg = pipe_reg.predict(X_test_reg)
rmse = root_mean_squared_error(y_test_reg, y_pred_reg)
print(f"Random Forest RMSE: {rmse:.4f}")
```

### 6.1. Otimização de Hiperparâmetros (Regressão)


```python
from sklearn.model_selection import GridSearchCV

param_grid_rf = {
    'regressor__n_estimators': [50, 100],
    'regressor__max_depth': [None, 10, 20]
}

best_rf_pipe = Pipeline(steps=[('preprocessor', preprocessor), ('regressor', RandomForestRegressor(random_state=42))])
grid_reg = GridSearchCV(best_rf_pipe, param_grid_rf, cv=5, scoring='neg_root_mean_squared_error')
grid_reg.fit(X_train_reg, y_train_reg)

print(f"Melhor RMSE (CV): {-grid_reg.best_score_:.4f}")
```

### 6.2. Importância das Variáveis (Regressão)


```python
importances = grid_reg.best_estimator_.named_steps['regressor'].feature_importances_
# Obter nomes das colunas após o preprocessor
ohe_cols = grid_reg.best_estimator_.named_steps['preprocessor'].named_transformers_['cat'].named_steps['onehot'].get_feature_names_out(categorical_features)
feature_names = numeric_features + list(ohe_cols)

feat_importances = pd.Series(importances, index=feature_names)
plt.figure(figsize=(10, 10))
feat_importances.nlargest(15).plot(kind='barh')
plt.title('Top 15 Variáveis Mais Importantes para MonthlyIncome')
plt.show()
```

## 7. Problema de Classificação: Prever Attrition

Utilizaremos o RandomForestClassifier para prever a saída de funcionários (Attrition).

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report
from imblearn.over_sampling import SMOTE
from imblearn.pipeline import Pipeline as ImbPipeline

X_clf = df_clean.drop(columns=[target_clf])
y_clf = df_clean[target_clf]

X_train_clf, X_test_clf, y_train_clf, y_test_clf = train_test_split(X_clf, y_clf, test_size=0.2, random_state=42, stratify=y_clf)

# Definir e treinar o modelo Random Forest Classifier com SMOTE
clf_final = ImbPipeline(steps=[
    ('preprocessor', preprocessor), 
    ('smote', SMOTE(random_state=42)),
    ('classifier', RandomForestClassifier(random_state=42))
])
clf_final.fit(X_train_clf, y_train_clf)

y_pred_clf = clf_final.predict(X_test_clf)
print("--- Random Forest (with SMOTE) ---")
print(classification_report(y_test_clf, y_pred_clf, digits=4))
```

## 8. Exportação dos Modelos Finais


### 7.1. Importância das Variáveis (Classificação)


```python
importances_clf = clf_final.named_steps['classifier'].feature_importances_
feat_importances_clf = pd.Series(importances_clf, index=feature_names)
plt.figure(figsize=(10, 10))
feat_importances_clf.nlargest(15).plot(kind='barh', color='orange')
plt.title('Top 15 Variáveis Mais Importantes para Attrition')
plt.show()
```
