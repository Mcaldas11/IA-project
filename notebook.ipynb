```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## 3. Análise Exploratória (EDA)

```python
# Distribuição de MonthlyIncome
sns.displot(df['MonthlyIncome'], kde=True, color='blue')
plt.title('Distribuição de MonthlyIncome')
plt.show()
```

```python
# Distribuição de Attrition
sns.countplot(x='Attrition', data=df, palette='viridis')
plt.title('Distribuição de Attrition')
plt.show()
```

```python
# Mapa de Correlação
plt.figure(figsize=(12, 8))
sns.heatmap(df.select_dtypes(include=[np.number]).corr(), annot=False, cmap='coolwarm')
plt.title('Matriz de Correlação (Numérica)')
plt.show()
```
